**Inline Question 2**

Although gradcheck is reliable softmax loss, it is possible that for SVM loss, once in a while, a dimension in the gradcheck will not match exactly. What could such a discrepancy be caused by? Is it a reason for concern? What is a simple example in one dimension where a svm loss gradient check could fail? How would change the margin affect of the frequency of this happening?

Note that SVM loss for a sample $(x_i, y_i)$ is defined as: $$L_i = \sum_{j\ne y_i}\max(0, s_j - s_{y_i} + \Delta)$$ where $j$ iterates over all classes except the correct class $y_i$ and $s_j$ denotes the classifier score for $j^{th}$ class. $\Delta$ is a scalar margin. For more information, refer to 'Multiclass Support Vector Machine loss' on [this](https://cs231n.github.io/linear-classify/) page.

*Hint: the SVM loss function is not strictly speaking differentiable.*


$\color{blue}{\textit Your Answer:}$ The SVM loss $\max(0, s_j - s_{y_i} + \Delta)$ is not differentiable at the kink where $s_j - s_{y_i} + \Delta = 0$: there the gradient is not uniquely defined (only a subgradient exists), so the analytic gradient (which commits to one branch) can disagree with the numerical gradient (which samples a small interval spanning the kink). This is not a concern because such points form a measure-zero set in the space of weights, so it is extremely unlikely to matter in practice; it is just a property of a non-smooth function. A one-dimensional example: with $\Delta = 1$, let the other-class score be $w$ and the correct-class score be 0, giving $L = \max(0, w - 1)$. At $w = 1$ the loss has a kink (value exactly 0) where the subgradient is ambiguous, so a gradcheck around $w = 1$ can mismatch. Increasing the margin $\Delta$ moves the kink to $s_j = s_{y_i} - \Delta$ and makes it more likely that some pair of scores falls exactly on (or very near) a kink during gradcheck, so such discrepancies become more frequent for larger $\Delta$.

**Inline Question 4** - *True or False*

Suppose the overall training loss is defined as the sum of the per-datapoint loss over all training examples. It is possible to add a new datapoint to a training set that would change the softmax loss, but leave the SVM loss unchanged.

$\color{blue}{\textit Your Answer:}$ **True**


$\color{blue}{\textit Your Explanation:}$ The softmax (cross-entropy) loss of a datapoint is $-\log(p_{y_i})$, which is always strictly positive because the probability of the correct class is always less than 1; so adding any new datapoint always increases (changes) the total softmax loss. In contrast, the SVM (hinge) loss of a new datapoint is exactly 0 whenever the correct class is classified with a margin larger than $\Delta$ (i.e. $s_{y_i} - s_j > \Delta$ for all $j \ne y_i$). Adding such a correctly-classified easy point therefore changes the softmax loss but leaves the SVM loss unchanged (W is held fixed; we only re-evaluate the sum of per-datapoint losses).
