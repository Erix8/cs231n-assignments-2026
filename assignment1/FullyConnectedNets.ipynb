## Inline Question 1:
Did you notice anything about the comparative difficulty of training the three-layer network vs. training the five-layer network? In particular, based on your experience, which network seemed more sensitive to the initialization scale? Why do you think that is the case?

## Answer:
The **five-layer** network was much more sensitive to the initialization scale. With weight_scale too small (e.g. 1e-5) the five-layer net barely learned (training accuracy stuck around 16%), while increasing it to 5e-2 let it overfit to 100%; the three-layer net, in contrast, overfit to 100% already with the standard 1e-2 scale. The reason is depth: a deeper network composes more linear transforms, so the scale of the forward activations and of the backward gradients is multiplied by roughly `weight_scale` at every layer and therefore grows/shrinks *exponentially* in the number of layers. Small differences in the initialization scale get amplified through many layers, which makes the five-layer network far more sensitive to weight_scale than the three-layer one.

## Inline Question 2:

AdaGrad, like Adam, is a per-parameter optimization method that uses the following update rule:

```
cache += dw**2
w += - learning_rate * dw / (np.sqrt(cache) + eps)
```

John notices that when he was training a network with AdaGrad that the updates became very small, and that his network was learning slowly. Using your knowledge of the AdaGrad update rule, why do you think the updates would become very small? Would Adam have the same issue?


## Answer:
AdaGrad's `cache` is a *cumulative* sum of squared gradients (`cache += dw**2`) that only ever grows and never decays. Since the effective step size is `learning_rate / (np.sqrt(cache) + eps)`, the denominator grows monotonically over time, so the updates shrink monotonically toward zero and the network eventually almost stops learning (parameters that receive frequent or large gradients get the smallest steps first). **Adam does not have this issue**: it maintains an exponentially decaying running average `v = beta2*v + (1-beta2)*dw**2` (with a bias-correction term), so `v` tracks the recent magnitude of the gradients instead of accumulating everything since the start of training. The step size therefore does not systematically shrink to zero and Adam keeps learning.
