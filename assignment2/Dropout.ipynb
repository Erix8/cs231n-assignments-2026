## Inline Question 1:
What happens if we do not divide the values being passed through inverse dropout by `p` in the dropout layer? Why does that happen?

## Answer:
If we drop neurons without dividing the surviving activations by `p`, the expected magnitude of the activations at test time no longer matches the magnitude seen during training. During training each activation is kept with probability `p` and otherwise set to zero, so the expected value of an output neuron is only `p * x`. At test time dropout is disabled and every neuron is active, so the same neuron outputs the full `x` -- which is `1/p` times larger than its training-time expectation.

That scale mismatch means the network effectively sees inputs (and produces activations) of a different scale at test time than it was trained on, so the learned weights are no longer matched to the activations and accuracy drops. Dividing the surviving activations by `p` (`out = x * mask / p`, i.e. "inverted dropout") rescales them so that the expected output during training equals `x` (because each activity is kept with probability `p` and scaled by `1/p`). Training and test time then have the same expected magnitude, so no rescaling of the weights or of the test-time activations is required.

# Regularization Experiment
As an experiment, we will train a pair of two-layer networks on 500 training examples: one will use no dropout, and one will use a keep probability of 0.25. We will then visualize the training and validation accuracies of the two networks over time.

```python
# Train two identical nets, one with dropout and one without.
np.random.seed(231)
num_train = 500
small_data = {
    'X_train': data['X_train'][:num_train],
    'y_train': data['y_train'][:num_train],
    'X_val': data['X_val'],
    'y_val': data['y_val'],
}

solvers = {}
dropout_choices = [1, 0.25]
for dropout_keep_ratio in dropout_choices:
    model = FullyConnectedNet(
        [500],
        dropout_keep_ratio=dropout_keep_ratio
    )
    print(dropout_keep_ratio)

    solver = Solver(
        model,
        small_data,
        num_epochs=25,
        batch_size=100,
        update_rule='adam',
        optim_config={'learning_rate': 5e-4,},
        verbose=True,
        print_every=100
    )
    solver.train()
    solvers[dropout_keep_ratio] = solver
    print()
```

```python
# Plot train and validation accuracies of the two models.
train_accs = []
val_accs = []
for dropout_keep_ratio in dropout_choices:
    solver = solvers[dropout_keep_ratio]
    train_accs.append(solver.train_acc_history[-1])
    val_accs.append(solver.val_acc_history[-1])

plt.subplot(3, 1, 1)
for dropout_keep_ratio in dropout_choices:
    plt.plot(
        solvers[dropout_keep_ratio].train_acc_history, 'o', label='%.2f dropout_keep_ratio' % dropout_keep_ratio)
plt.title('Train accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend(ncol=2, loc='lower right')

plt.subplot(3, 1, 2)
for dropout_keep_ratio in dropout_choices:
    plt.plot(
        solvers[dropout_keep_ratio].val_acc_history, 'o', label='%.2f dropout_keep_ratio' % dropout_keep_ratio)
plt.title('Val accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend(ncol=2, loc='lower right')

plt.gcf().set_size_inches(15, 15)
plt.show()
```

## Inline Question 2:
Compare the validation and training accuracies with and without dropout -- what do your results suggest about dropout as a regularizer?

## Answer:
Without dropout the network fits the tiny training set extremely well -- training accuracy climbs to ~0.92 (peaking around 0.98) -- while validation accuracy stalls around 0.26-0.31, so the train/validation gap is large. That is the classic signature of overfitting: the model memorizes the 500 training examples instead of learning a generalizable function.

With dropout (keep probability 0.25) the final training accuracy is slightly lower (~0.90), but the best validation accuracy is higher (peaking around 0.34 vs ~0.31) and the final validation accuracy is clearly better (~0.31 vs ~0.26). Because dropout randomly zeroes activations during training, the network cannot rely on any single unit or co-adapted group of units, so it is forced to learn redundant, more robust features -- effectively training an ensemble of subnetworks that share weights. The result is better generalization on the held-out set at the cost of a small amount of training-set accuracy. This supports dropout as an effective regularizer: it trades a little training fit for a smaller train/validation generalization gap.
